```julia
using NeuralParam
using SpeedyWeather, AnalyticBandRadiation
SpeedyExt = Base.get_extension(AnalyticBandRadiation,
                                     :AnalyticBandRadiationSpeedyWeatherExt)
using Plots
```

```julia
TRUNC = 31
NLAYERS = 8
SG = SpectralGrid(trunc=TRUNC, nlayers=NLAYERS)
```

```julia
schemes = merge(
    (;  FreeRun = nothing,
        OBLW    = OneBandLongwave(SG; transmissivity = FriersonLongwaveTransmissivity(SG)),
        ABRLW   = SpeedyExt.SpeedyAnalyticBandLongwave(SG)),
        collect_schemes(["CLLW_default"]),
)
```

```julia
results = evaluate_benchmark(schemes; spectral_grid=SG)
print_benchmark(results)
```
